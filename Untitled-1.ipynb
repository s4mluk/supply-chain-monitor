```python
import yfinance as yf

def get_market_indicators():
    tickers = {
        "Baltic Dry Index": "BDRY",   # tai BDI ETF: "BDRY"
        "Crude Oil (WTI)":  "CL=F",
        "Copper":           "HG=F"
    }
    
    results = []
    for name, ticker in tickers.items():
        try:
            t = yf.Ticker(ticker)
            hist = t.history(period="5d")
            
            if hist.empty:
                continue
                
            latest = hist["Close"].iloc[-1]
            prev   = hist["Close"].iloc[0]
            change = ((latest - prev) / prev) * 100
            
            results.append({
                "indicator": name,
                "price":     round(float(latest), 2),
                "change_5d": round(float(change), 2)
            })
        except Exception as e:
            print(f"Virhe {name}: {e}")
    
    return results

# Testataan
if __name__ == "__main__":
    indicators = get_market_indicators()
    for item in indicators:
        print(item)
```

```python
import requests

def get_port_weather():
    ports = {
        "Singapore":   (1.29,  103.85),
        "Rotterdam":   (51.96,  4.13),
        "Shanghai":    (31.23, 121.47),
        "Los Angeles": (33.74, -118.27)
    }
    
    results = {}
    for port, (lat, lon) in ports.items():
        try:
            r = requests.get(
                "https://api.open-meteo.com/v1/forecast",
                params={
                    "latitude":       lat,
                    "longitude":      lon,
                    "current":        "wind_speed_10m,weather_code",
                    "wind_speed_unit": "ms"
                }
            ).json()
            
            wind = r["current"]["wind_speed_10m"]
            
            results[port] = {
                "wind_speed_ms": wind,
                "alert":         wind > 15
            }
        except Exception as e:
            print(f"Virhe {port}: {e}")
    
    return results

# Testataan
if __name__ == "__main__":
    weather = get_port_weather()
    for port, data in weather.items():
        alert = "⚠️" if data["alert"] else "✓"
        print(f"{alert} {port}: tuuli {data['wind_speed_ms']} m/s")
```

```python
import lseg.data as ld
import yfinance as yf
import requests
from datetime import timedelta

ld.open_session()

# ================================
# 1. UUTISET — Refinitiv
# ================================
def get_news():
    queries = [
        # 1. Aasian reitit — tärkein consumer goods -yrityksille
        "Asia shipping OR China port OR Suez Canal OR Red Sea disruption",
        # 2. Raaka-aineet ja logistiikka
        "freight rates OR container shortage OR supply chain disruption",
        # 3. Consumer goods -toimiala
        "consumer goods supply chain OR FMCG logistics OR retail inventory"
    ]
    
    all_results = []
    seen_ids = set()
    
    for query in queries:
        try:
            headlines = ld.news.get_headlines(
                query=query,
                start="2024-01-01",
                end=timedelta(days=0),
                count=8
            )
            
            for idx, row in headlines.iterrows():
                # suodatetaan Twitter/X pois
                if row["sourceCode"] in ["NS:TWIT", "NS:TWTR"]:
                    continue
                if row["headline"].startswith("@"):
                    continue
                # estetään duplikaatit
                if row["storyId"] in seen_ids:
                    continue
                
                seen_ids.add(row["storyId"])
                all_results.append({
                    "headline": row["headline"],
                    "date":     str(idx),
                    "storyId":  row["storyId"],
                    "category": query.split(" OR ")[0]
                })
        except Exception as e:
            print(f"Virhe haussa '{query}': {e}")
    
    return all_results[:20]

# ================================
# 2. MARKKINAINDIKAATTORIT — yfinance
# ================================
def get_market_indicators():
    tickers = {
        "Freight ETF (BDI proxy)": "BDRY",
        "Crude Oil (WTI)":          "CL=F",
        "Copper":                   "HG=F"
    }
    
    results = []
    for name, ticker in tickers.items():
        try:
            hist = yf.Ticker(ticker).history(period="5d")
            
            if hist.empty:
                print(f"Ei dataa: {name}")
                continue
            
            latest = hist["Close"].iloc[-1]
            prev   = hist["Close"].iloc[0]
            change = ((latest - prev) / prev) * 100
            
            results.append({
                "indicator": name,
                "price":     round(float(latest), 2),
                "change_5d": round(float(change), 2)
            })
        except Exception as e:
            print(f"Virhe {name}: {e}")
    
    return results

# ================================
# 3. SATAMASÄÄ — Open-Meteo
# ================================
def get_port_weather():
    ports = {
        "Singapore":   (1.29,  103.85),
        "Rotterdam":   (51.96,  4.13),
        "Shanghai":    (31.23, 121.47),
        "Los Angeles": (33.74, -118.27)
    }
    
    results = {}
    for port, (lat, lon) in ports.items():
        try:
            r = requests.get(
                "https://api.open-meteo.com/v1/forecast",
                params={
                    "latitude":        lat,
                    "longitude":       lon,
                    "current":         "wind_speed_10m,weather_code",
                    "wind_speed_unit": "ms"
                }
            ).json()
            
            wind = r["current"]["wind_speed_10m"]
            results[port] = {
                "wind_speed_ms": wind,
                "alert":         wind > 15
            }
        except Exception as e:
            print(f"Virhe {port}: {e}")
    
    return results

# ================================
# TESTI — aja kaikki kerralla
# ================================
if __name__ == "__main__":
    print("=== UUTISET ===")
    news = get_news()
    for item in news:
        print(f"[{item['category']}]")
        print(f"  {item['date']} — {item['headline']}")
    
    print(f"\nYhteensä {len(news)} uutista\n")
    
    print("=== MARKKINAT ===")
    indicators = get_market_indicators()
    for item in indicators:
        print(f"{item['indicator']}: {item['price']} ({item['change_5d']:+.1f}% 5pv)")
    
    print("\n=== SATAMASÄÄ ===")
    weather = get_port_weather()
    for port, data in weather.items():
        alert = "⚠️" if data["alert"] else "✓"
        print(f"{alert} {port}: {data['wind_speed_ms']} m/s")
```
